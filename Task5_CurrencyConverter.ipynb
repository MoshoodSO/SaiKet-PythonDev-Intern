```python
import requests
```

```python
def convert_currency(amount, from_currency, to_currency):
    """Convert currency using the Frankfurter API."""

    url = "https://api.frankfurter.app/latest"

    params = {
        "amount": amount,
        "from": from_currency,
        "to": to_currency
    }

    response = requests.get(url, params=params, timeout=10)

    if response.status_code != 200:
        raise Exception("Unable to get exchange rate.")

    
    data = response.json()


    return data["rates"][to_currency]
```

```python
def main():
    print("=" * 45)
    print("          CURRENCY CONVERTER")
    print("=" * 45)

    currencies = {
                  "AUD": "Australian Dollar",
                  "BRL": "Brazilian Real",
                  "CAD": "Canadian Dollar",
                  "CHF": "Swiss Franc",
                  "CNY": "Chinese Yuan",
                  "CZK": "Czech Koruna",
                  "DKK": "Danish Krone",
                  "GBP": "British Pound",
                  "HKD": "Hong Kong Dollar",
                  "HUF": "Hungarian Forint",
                  "IDR": "Indonesian Rupiah",
                  "ILS": "Israeli New Shekel",
                  "INR": "Indian Rupee",
                  "ISK": "Icelandic Króna",
                  "JPY": "Japanese Yen",
                  "KRW": "South Korean Won",
                  "MXN": "Mexican Peso",
                  "MYR": "Malaysian Ringgit",
                  "NOK": "Norwegian Krone",
                  "NZD": "New Zealand Dollar",
                  "PHP": "Philippine Peso",
                  "PLN": "Polish Zloty",
                  "RON": "Romanian Leu",
                  "SEK": "Swedish Krona",
                  "SGD": "Singapore Dollar",
                  "THB": "Thai Baht",
                  "TRY": "Turkish Lira",
                  "USD": "United States Dollar",
                  "ZAR": "South African Rand"
                }

    
    print("\nAvailable currencies:")

    for code, name in currencies.items():
        print(f"{code} - {name}")

    try:
        # Get amount
        amount = float(input("\nEnter amount: "))

        if amount <= 0:
            print("Error: Amount must be greater than zero.")
            return

        # Get currencies
        from_currency = input("Enter currency to convert from: ").upper().strip()

        to_currency = input("Enter currency to convert to: ").upper().strip()

        # Validate currencies
        if from_currency not in currencies:
            print("Error: Invalid source currency.")
            return

        if to_currency not in currencies:
            print("Error: Invalid target currency.")
            return

        # Same currency
        if from_currency == to_currency:
            result = amount

        else:
            result = convert_currency(
                amount,
                from_currency,
                to_currency
            )

        print("\n" + "=" * 45)
        print(
            f"{amount:,.2f} {from_currency} = "
            f"{result:,.2f} {to_currency}"
        )
        print("=" * 45)

    except ValueError:
        print("Error: Please enter a valid number.")

    except requests.exceptions.ConnectionError:
        print("Error: No internet connection.")

    except requests.exceptions.Timeout:
        print("Error: Request timed out.")

    except Exception as error:
        print(f"Error: {error}")


if __name__ == "__main__":
    main()
```
