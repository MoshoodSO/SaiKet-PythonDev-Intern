```python
from collections import Counter
import pandas as pd
```

```python
def WordCounter(file, top_n = 5):
    '''
    A function that count the number of words in a file and return the n most occuring word.

    This function open a file and returns the the n most occuring word in the file with its frequency.

    Args:
        file (str): The directory of the file
        top_n (int): The number pf the most occuring word. (5)

    Return:
        list: the top n words and its frequency.

    Raise:
        FileNotFoundError: If the file is not found or the directoryis not correct.

    Example:
    >>> WordCounter("docs.txt")
        	words	counts
        0	the     5
        1	and	    3
        2	a	    2
        3	program	2
        4	user	2

    '''
    
    try:
        word_counter = {}
        with open(file, 'r') as f:
            content = f.read()

        # the word counter
        content_list = content.split()
        return pd.DataFrame(Counter(content_list).most_common(top_n), columns = ["words", "counts"])

    except FileNotFoundError:
        return f"No such file or directory: {file}"
```
