```python
def Replace_word(file, word, replacement):
    '''
    A function that replace a word with another.

    This function open a file and replace a given word with a given replacement the returns the "Successful" after imputing it in another file.

    Args:
        file (str): The directory of the file.
        word (str): Word to find and be replaced.
        replacement (str): Word to imput (replace).

    Return:
        str: the new sentences.

    Raise:
        FileNotFoundError: If the file is not found or the directoryis not correct.

    Example:
    >>> Replace_word("docs.txt","number", "mine")
    'Description: Write a program that randomly generates a mine between 1 and 100. The user has to guess the mine, 
    and the program will give feedback if the guess is too high or too low. \nCreate functions for each operation.
    \nTake two inputs from the user and allow them to select the desired operation.\nHandle division by zero with 
    appropriate error messages.'
    

    '''
    
    try:
        with open(file, 'r') as f:
            content = f.read()

        #replacement
        content = content.replace(word, replacement)

        # rewriting
        with open("Editted_file.txt", 'w') as file:
            file.write(content)

        return f"Successful! File inputed in 'Editted_file.txt'"

        
        # content

    except FileNotFoundError:
        return f"No such file or directory: {file}"
    
```

```python
Replace_word("docs.txt","number", "mine")
```
