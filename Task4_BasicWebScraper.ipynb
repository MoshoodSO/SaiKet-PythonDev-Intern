```python
## Required Libraries
import requests
from bs4 import BeautifulSoup
import csv
import pandas as pd
```

```python
# Website to scrape
url = "https://news.ycombinator.com/"

# Send a request to retrieve the webpage
headers = {"User-Agent": "Mozilla/5.0"}

response = requests.get(url, headers=headers, timeout=10)
```

```python
# Check if the request was successful
if response.status_code == 200:
    print("Webpage retrieved successfully.")

    # Parse the HTML
    soup = BeautifulSoup(response.text, "html.parser")

    # Find news headlines
    headlines = soup.select(".titleline > a")

    # Store extracted data
    news_data = []

    for headline in headlines:
        title = headline.get_text(strip=True)
        link = headline.get("href")

        news_data.append({
            "title": title,
            "link": link
        })

    # Save the data to a CSV file
    with open("news_headlines.csv", "w", newline="", encoding="utf-8") as file:
        writer = csv.DictWriter(file, fieldnames=["title", "link"])

        writer.writeheader()
        writer.writerows(news_data)

    print(f"{len(news_data)} headlines saved to news_headlines.csv")

else:
    print(f"Failed to retrieve webpage. Status code: {response.status_code}")
```
