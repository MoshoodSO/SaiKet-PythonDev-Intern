```python
# import library
import random
```

```python
def number_guessing_game():
    """
    Run a number guessing game loop against a secret target.

    This function manages the interactive user input loop, checks guesses against the target number, and provides feedback on whether the guess 
    was too high or too low.

    Args:
        secret_number (int): The hidden number the player must guess.
        
    Returns:
        bool: True if the player guesses correctly, False if they run out of turns.

    Raises:
        ValueError: If the number of attempt are exhausted.

    """
    # The number to guess
    num = random.randint(1,100)

    ## Imput for the user's guess
    guess = int(input("Enter your guess:"))

    # Input Counter
    i = 0
    while guess != num:
        if i == 10:
            print("Time out!!")
            break
        elif guess < num:
            print(f" ❌ Too low! You only have {10 - i} trial(s) left")
        elif guess > num:
            print(f" ❌ Too high! You only have {10 - i} trial(s) left")
        guess = int(input("Enter another guess:"))
        i += 1
    print(" ✅ Correct guess!!")
```

```python
number_guessing_game() 
```
